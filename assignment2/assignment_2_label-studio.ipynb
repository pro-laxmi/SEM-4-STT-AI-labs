```python
import requests
import json

class LabelStudioClient:
    def __init__(self, url='http://localhost:8080', api_key=None):
        self.url = url
        self.headers = {
            'Authorization': f'Bearer {api_key}',
            'Content-Type': 'application/json'
        }

    def create_project(self, title, config):
        response = requests.post(
            f"{self.url}/api/projects/",
            json={"title": title, "label_config": config},
            headers=self.headers
        )
        response.raise_for_status()
        return response.json()

    def import_tasks(self, project_id, tasks):
        response = requests.post(
            f"{self.url}/api/projects/{project_id}/import",
            json=tasks,
            headers=self.headers
        )
        response.raise_for_status()
        return response.json()

    def export_annotations(self, project_id):
        response = requests.get(
            f"{self.url}/api/projects/{project_id}/export",
            headers=self.headers
        )
        response.raise_for_status()
        return response.json()
```
